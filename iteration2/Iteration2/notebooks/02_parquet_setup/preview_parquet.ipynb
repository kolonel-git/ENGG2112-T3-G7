# Preview processed turbine Parquet data
Edit the year, turbine ID and preview row count below. Paths resolve automatically from the project or its notebook folders.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display

# Change these settings to preview another turbine or year.
YEAR = 2016
TURBINE_ID = '2304511'
ROWS = 20

# Find Iteration2 from either the repository root or a notebook subfolder.
ROOT = next((candidate for parent in (Path.cwd(), *Path.cwd().parents)
             for candidate in (parent, parent / 'iteration2' / 'Iteration2')
             if (candidate / 'Data_Parquets_10min').is_dir()), None)
if ROOT is None:
    raise FileNotFoundError('Cannot find Data_Parquets_10min. Run from this project and complete preprocessing first.')

PARQUET_PATH = ROOT / 'Data_Parquets_10min' / str(YEAR) / f'Station_{TURBINE_ID}' / 'data.parquet'
if not PARQUET_PATH.is_file():
    raise FileNotFoundError(f'No processed file at {PARQUET_PATH}. Check YEAR and TURBINE_ID, or run preprocessing for this year.')

print(f'Previewing: {PARQUET_PATH}')
df = pd.read_parquet(PARQUET_PATH)
display(df.head(ROWS))
print(f'{len(df):,} rows, {len(df.columns)} columns')
display(pd.DataFrame({'dtype': df.dtypes.astype(str), 'missing_values': df.isna().sum()}))
```
